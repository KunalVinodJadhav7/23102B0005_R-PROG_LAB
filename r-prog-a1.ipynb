```r
# Beijing Multi-Site Air Quality Data Cleaning in R

# Load Libraries

library(tidyverse)

# Task 1: Import and Inspect the Dataset

dataset_folder <- "../input/datasets/sid321axn/beijing-multisite-airquality-data-set"

files <- list.files(
  path = dataset_folder,
  pattern = "\\.csv$",
  full.names = TRUE
)

cat("Number of CSV files:", length(files), "\n\n")

# Read all CSV files and combine them
air_quality <- map_dfr(files, read_csv, show_col_types = FALSE)

cat("Dataset Dimensions:\n")
print(dim(air_quality))

cat("\nStructure of Dataset:\n")
str(air_quality)

cat("\nSummary Statistics:\n")
summary(air_quality)

cat("\nFirst Six Rows:\n")
head(air_quality)

# Task 2: Understand NA, NULL and NaN

cat("\n========== NA, NULL and NaN ==========\n")

value1 <- NA
value2 <- NULL
value3 <- 0/0

cat("Is value1 NA? ", is.na(value1), "\n")
cat("Is value2 NULL? ", is.null(value2), "\n")
cat("Is value3 NaN? ", is.nan(value3), "\n")

# Task 3: Create Missing Value Summary Function

missing_summary <- function(data){

  data.frame(

    Variable = names(data),

    Missing_Values = sapply(data,function(x) sum(is.na(x))),

    Missing_Percentage =
      round(sapply(data,function(x) mean(is.na(x))*100),2)

  )

}

missing_before <- missing_summary(air_quality)

cat("\nMissing Value Summary:\n")
print(missing_before)

write_csv(missing_before,"missing_summary.csv")

# Task 4: Identify Invalid Numerical Results

numeric_columns <- names(air_quality)[sapply(air_quality,is.numeric)]

invalid_values <- data.frame(

  Variable = numeric_columns,

  NaN_Count =
    sapply(air_quality[numeric_columns],function(x) sum(is.nan(x))),

  Infinite_Count =
    sapply(air_quality[numeric_columns],function(x) sum(is.infinite(x)))

)

cat("\nInvalid Numerical Values:\n")
print(invalid_values)

# Task 5: Handle Missing Numerical Values Using a Loop

clean_data <- air_quality

for(column in numeric_columns){

  median_value <- median(clean_data[[column]],na.rm=TRUE)

  clean_data[[column]][is.na(clean_data[[column]])] <- median_value

}

# Task 6: Handle Missing Categorical Values

mode_function <- function(x){

  values <- unique(na.omit(x))

  values[which.max(tabulate(match(x,values)))]

}

categorical_columns <- names(clean_data)[sapply(clean_data,is.character)]

for(column in categorical_columns){

  mode_value <- mode_function(clean_data[[column]])

  clean_data[[column]][is.na(clean_data[[column]])] <- mode_value

}

# Task 7: Implement Error Handling

safe_import <- function(path){

  tryCatch(

    {

      read_csv(path,show_col_types = FALSE)

    },

    error = function(e){

      cat("Error while reading file:\n")

      print(e$message)

      return(NULL)

    }

  )

}

test_dataset <- safe_import(files[1])

# Task 8: Compare Missing Values Before and After Cleaning

missing_after <- missing_summary(clean_data)

comparison_table <- data.frame(

  Variable = missing_before$Variable,

  Missing_Before = missing_before$Missing_Values,

  Missing_After = missing_after$Missing_Values

)

cat("\nComparison Table:\n")
print(comparison_table)

write_csv(comparison_table,"missing_comparison.csv")

# Task 9: Generate One Visualization

plot <- ggplot(

  missing_before,

  aes(

    x = reorder(Variable,Missing_Percentage),

    y = Missing_Percentage

  )

) +

geom_col(fill="steelblue") +

coord_flip() +

labs(

title="Missing Values Before Cleaning",

x="Variables",

y="Percentage of Missing Values"

) +

theme_minimal()

print(plot)

ggsave(

"missing_values_plot.png",

plot,

width=10,

height=6,

dpi=300

)

# Task 10: Export the Cleaned Dataset

write_csv(clean_data,"cleaned_air_quality.csv")

cat("\nCleaned dataset exported successfully.\n")

# Display Final Results

cat("\n==============================\n")
cat("FILES CREATED\n")
cat("==============================\n")

cat("1. missing_summary.csv\n")
cat("2. missing_comparison.csv\n")
cat("3. missing_values_plot.png\n")
cat("4. cleaned_air_quality.csv\n")

cat("\nCleaning Completed Successfully.\n")

# Interpretation (100–150 words)

cat("

INTERPRETATION

The Beijing Multi-Site Air Quality dataset from all monitoring stations was
combined into a single dataset and examined for missing values. A custom
function summarized missing observations and identified variables with
incomplete data. Numerical missing values were replaced using the median
through a loop, while categorical missing values were imputed using the mode.
The dataset was also checked for invalid numerical values such as NaN and
infinite values. Error handling was implemented using tryCatch() to ensure
robust file importing. A comparison of missing values before and after cleaning
confirmed that all missing observations were successfully handled. The
visualization clearly highlighted the variables with the highest percentage of
missing data before cleaning. The cleaned dataset is now complete and suitable
for further statistical analysis, visualization, and predictive modeling.

")
```
